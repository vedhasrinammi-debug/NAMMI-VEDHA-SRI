**Example C2 — The Grading Weight System**
Design your own weighting scheme for your current semester subjects.
Your code must answer:
(i) How does the weighted average differ from the simple unweighted average?
(ii) Which student benefits most from the weighting? Which is disadvantaged?
(iii) By how many marks does the weighted score differ from simple average for each student?

```python
import numpy as np
marks = np.array([[85, 90, 75, 80, 95],[70, 80, 90, 75, 85]])
weights = np.zeros(5) + 1.0
weights[0] = 1.5
weights[4] = 2.0
unweighted_avg = np.mean(marks, axis=1)
weighted_avg = np.average(marks, axis=1, weights=weights)
diff = weighted_avg - unweighted_avg
print("(i) Group average difference:", np.mean(weighted_avg) - np.mean(unweighted_avg))
print("(ii) Benefited student index:", np.argmax(diff), "| Disadvantaged student index:", np.argmin(diff))
print("(iii) Individual differences:", diff)
```

**Example D3 — The What-If Explorer**

What happens when you simulate 10,000 flips instead?
Your code must answer:
(i) What did your simulation predict for the defined question?
(ii) Run the simulation twice with different seeds. How much does the answer vary?
(iii) Write a one-line comment in your code stating your conclusion.

```python
import numpy as np
def run_sim(seed_value):
    np.random.seed(seed_value)
    dice1 = np.random.randint(1, 7, 10000)
    dice2 = np.random.randint(1, 7, 10000)
    sums = dice1 + dice2
    successes = np.sum((sums == 7) | (sums == 11))
    return successes / 10000
prob_seed1 = run_sim(42)
prob_seed2 = run_sim(123)
print(f"(i) Predicted probability (Seed 42): {prob_seed1 * 100:.2f}%")
variance = abs(prob_seed1 - prob_seed2) * 100
print(f"(ii) Probability with Seed 123: {prob_seed2 * 100:.2f}% | Absolute variance: {variance:.2f}%")
```

**Example G3 — The Discovery**
At what distance does travel time at 60 kmph save more than 1 hour vs 40 kmph?
Your code must answer:
(i) Identify the exact threshold or crossover point from the array.
(ii) State your finding as a one-line conclusion written as a code comment.
(iii) Write which function (arange or linspace) you used and why in a comment.

```python
import numpy as np
distances = np.arange(10, 201, 10)
time_40 = distances / 40
time_60 = distances / 60
time_saved = time_40 - time_60
threshold_distance = distances[time_saved > 1][0]
print("(i) Threshold crossover distance:", threshold_distance, "km")

# (ii) Conclusion: Traveling at 60 kmph saves more than 1 hour compared to 40 kmph once the distance exceeds 120 km.
# (iii) Selection: Used 'arange' to define the sequence by a specific, easy-to-read step size of 10 km intervals.
```
